```python
# QuantUI Viewer - browser build (Pyodide). Packages come from this site's
# bundled wheels; QuantUI itself is installed without its server-side deps
# (Jupyter, Voila, kaleido), which a browser kernel neither has nor needs.
import piplite

await piplite.install(
    ["numpy", "matplotlib", "requests", "plotly", "narwhals", "py3dmol", "ipywidgets"]
)
await piplite.install("quantui", deps=False)

import os

os.environ.setdefault("QUANTUI_RESULTS_DIR", "sample_results")
from quantui.app import QuantUIApp

QuantUIApp(viewer=True).display()
```
