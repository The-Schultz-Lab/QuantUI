## Step 1: Setup

First, we load QuantUI.

### What's happening here?
- Loading the QuantUI modules
- Checking that PySCF (the quantum chemistry engine) is available

Every calculation in this tutorial runs **inside this notebook's Python
session** — nothing is sent anywhere else.

### Run the cell below ⬇️

```python
# Load QuantUI
import io

from quantui import Molecule, parse_xyz_input

try:
    from quantui import run_in_session

    PYSCF_AVAILABLE = True
    print("✓ QuantUI and PySCF loaded — ready for your first calculation.")
except ImportError:
    PYSCF_AVAILABLE = False
    print("PySCF not available — calculations will not run.")
    print("PySCF requires Linux, macOS, or WSL.")
```

## Step 2: Define Your Molecule - Water (H₂O)

We'll start with water - one of the most important molecules in chemistry!

### About Water (H₂O)
- **Chemical formula:** H₂O
- **Atoms:** 1 oxygen + 2 hydrogen
- **Shape:** Bent (not linear!)
- **Bond angle:** ~104.5°
- **Why it matters:** Essential for life, unique properties

### XYZ Coordinates Explained
Each line has: `ATOM  X  Y  Z`
- **ATOM**: Element symbol (O, H, C, etc.)
- **X, Y, Z**: Position in 3D space (in Ångstroms)
- (0,0,0) is arbitrary origin point

### Run the cell below to define water ⬇️

```python
# Define water molecule coordinates
water_xyz = """O  0.000  0.000  0.000
H  0.757  0.587  0.000
H -0.757  0.587  0.000"""

# Molecular properties
charge = 0          # Neutral molecule (no net charge)
multiplicity = 1    # Singlet state (all electrons paired)

print("Water Molecule (H₂O)")
print("="*40)
print(water_xyz)
print("="*40)
print(f"Charge: {charge}")
print(f"Multiplicity: {multiplicity}")
print("\n💡 Tip: Multiplicity = 1 means all electrons are paired")
```

### ✅ Checkpoint 1

**Before proceeding, make sure:**
- [ ] You see 3 lines of coordinates (O, H, H)
- [ ] Charge is 0 (neutral)
- [ ] Multiplicity is 1 (paired electrons)

**Questions to think about:**

2. What would happen if we used the wrong charge?

---

## Step 3: Create the Molecule Object

Now we'll create a QuantUI `Molecule` object that validates our input.

### What's being validated?
- Atom symbols are valid elements
- Coordinates are numbers
- Charge and multiplicity are compatible with the number of electrons

### Run the cell below ⬇️

```python
# Parse coordinates and create molecule
atoms, coordinates = parse_xyz_input(water_xyz)

current_molecule = Molecule(
    atoms=atoms,
    coordinates=coordinates,
    charge=charge,
    multiplicity=multiplicity,
)

print("✓ Molecule created successfully!")
print("\n📊 Molecule Summary:")
print(f"   Formula: {current_molecule.get_formula()}")
print(f"   Total atoms: {len(atoms)}")
print(f"   Total electrons: {current_molecule.get_electron_count()}")
print(f"   Charge: {charge}")
print(f"   Multiplicity: {multiplicity}")

# Optional 3D view (needs py3Dmol, which QuantUI installs)
try:
    from quantui import PY3DMOL_AVAILABLE, display_molecule
except ImportError:
    PY3DMOL_AVAILABLE = False
if PY3DMOL_AVAILABLE:
    print("\n🔬 3D Visualization:")
    display_molecule(current_molecule, style="ball+stick", show_info=False)
else:
    print("\n💡 Install py3Dmol to see a 3D view")
```

### ✅ Checkpoint 2

**Verify your molecule:**
- Formula should be: **H2O**
- Total atoms: **3**
- Total electrons: **10** (8 from O, 1 each from H)
- If you see 3D visualization, molecule should look bent, not linear

**Understanding electrons:**
- Oxygen: 8 electrons (atomic number = 8)
- Hydrogen: 1 electron each (atomic number = 1)
- Total: 8 + 1 + 1 = 10 electrons

## Step 4: Configure the Calculation

Now we'll set up the quantum chemistry calculation.

### Method: RHF (Restricted Hartree-Fock)
- **RHF** = Restricted Hartree-Fock
- Good for closed-shell molecules (all electrons paired)
- Treats electron-electron interactions approximately
- Fast and reliable for small molecules

### Basis Set: 6-31G
- **Basis set** = Mathematical functions to describe electrons
- **6-31G** = Medium-quality basis (good balance of speed/accuracy)
- Larger basis = more accurate but slower
- Common choices: STO-3G (small), 6-31G (medium), cc-pVDZ (large)

### Run the cell below ⬇️

```python
# Calculation settings
method = "RHF"  # Hartree-Fock method
basis = "6-31G"  # Basis set

print("⚙️  Calculation Configuration")
print("=" * 50)
print(f"Method: {method}")
print(f"Basis Set: {basis}")
print("=" * 50)

print("\n📖 What is Hartree-Fock?")
print("   • Quantum method for calculating electronic structure")
print("   • Solves the Schrödinger equation approximately")
print("   • Gives the total energy and molecular orbitals")
print("   • 'Self-Consistent Field' (SCF): an iterative approach")

print("\n💡 What to expect:")
print("   • Energy: about -75.98 Hartree for this H₂O geometry with RHF/6-31G")
print("   • SCF iterations: about 8")
print("   • Runtime: about a second on a laptop")
```

### ✅ Checkpoint 3

**Review your settings:**
- Method: **RHF** (good for H₂O, a closed-shell molecule)
- Basis: **6-31G** (medium quality)

**Key concept - What is a Hartree?**
- The unit of energy in quantum chemistry ("atomic units")
- Named after Douglas Hartree
- 1 Hartree = 27.211 eV = 627.5 kcal/mol
- The total energy (about -76 Ha for water) is measured relative to all the
  nuclei and electrons infinitely far apart and at rest. It is negative because
  the molecule is far more stable than that collection of separated particles.

---

## Step 5: Run the Calculation! 🚀

Time to run your first quantum chemistry calculation.

### What happens when you run it?
1. QuantUI builds the PySCF molecule from your `Molecule` object
2. PySCF starts from an initial guess for the orbitals
3. It repeats the SCF cycle — build the orbitals, recompute the electron
   repulsion, rebuild — until nothing changes any more (**convergence**)
4. QuantUI collects the results (energy, orbitals, charges, dipole) into a
   `SessionResult`

`warm_start=False` makes PySCF start from scratch so you see the real number of
SCF iterations. (Normally QuantUI reuses the density from an earlier run of the
same molecule as the starting guess, so a repeat run can converge in one
iteration.)

### Run the cell below ⬇️

```python
# Run RHF/6-31G on water in this Python session
log = io.StringIO()  # PySCF's detailed output goes here

if not PYSCF_AVAILABLE:
    result = None
    print("Skipping — PySCF not available.")
else:
    result = run_in_session(
        current_molecule,
        method=method,
        basis=basis,
        progress_stream=log,
        warm_start=False,
    )
    print("✓ Calculation finished.")
    print("  (Run print(log.getvalue()) in a new cell to see PySCF's full output.)")
```

### ✅ Checkpoint 4

**The calculation finished if you see:**
- ✅ "Calculation finished."
- ✅ No error message

---

## Step 6: View Results! 🎉

Let's see what we learned about water.

### What results to look for:
1. **Total Energy** - How stable is the molecule?
2. **SCF Convergence** - Did the calculation succeed?
3. **Number of iterations** - How hard was it?
4. **HOMO-LUMO gap** - The energy gap between the highest occupied and lowest
   unoccupied molecular orbitals
5. **Dipole moment and Mulliken charges** - How the electrons are distributed

### Run the cell below ⬇️

```python
# View calculation results
if result is None:
    print("No result — PySCF was not available.")
else:
    print("📊 CALCULATION RESULTS")
    print("=" * 60)
    print(f"Molecule: {result.formula}    Method: {result.method}/{result.basis}")
    print(f"Total energy:      {result.energy_hartree:.6f} Hartree")
    print(f"                   {result.energy_hartree * 627.5095:.1f} kcal/mol")
    print(f"SCF converged:     {result.converged}")
    print(f"SCF iterations:    {result.n_iterations}")
    print(f"HOMO-LUMO gap:     {result.homo_lumo_gap_ev:.2f} eV")
    print(f"Dipole moment:     {result.dipole_moment_debye:.2f} Debye")
    print("Mulliken charges:")
    for symbol, q in zip(result.atom_symbols, result.mulliken_charges):
        print(f"   {symbol}: {q:+.3f}")
    print("=" * 60)

    expected = -75.98
    if abs(result.energy_hartree - expected) < 0.01:
        print(f"✅ Matches the expected value (about {expected:.2f} Ha)")
    else:
        print(f"⚠️  Different from the expected {expected:.2f} Ha — check your inputs")
```

### ✅ Final Checkpoint

**Success criteria:**
- ✅ Energy about **-75.98 Hartree**
- ✅ SCF converged: **True**
- ✅ About **8** SCF iterations
- ✅ HOMO-LUMO gap about **19 eV**, dipole about **2.6 D**
- ✅ Oxygen carries a negative Mulliken charge (about -0.79); each hydrogen a
  positive one (about +0.40)

**Understanding your results:**
- **Energy (-75.98 Ha)**: total electronic energy of H₂O (plus nuclear repulsion)
  at this geometry
- **SCF convergence**: the iterative method found a self-consistent solution
- **Charges and dipole**: oxygen is more electronegative, so it pulls electron
  density from the hydrogens — this is why water is polar. (The experimental
  dipole of water is 1.85 D; Hartree-Fock with a small basis overestimates it.)
